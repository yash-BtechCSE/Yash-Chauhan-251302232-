```python
class Node:                     # Node CLass
  def __init__(self, data):     # Constructor
    self.data = data            # CLass Attribute to store Data
    self.next = None            # CLass Attribute to store Address/Reference of next node - NULL


class Operations:
  def __init__(self):
    self.head = None       # Engine - Starting Point    #Reference
    self.tail = None       # Guard Room - Ending Point  #Reference
    print("Linked List Operations Object Created with head=Null and Tail=Null, NO NODES ARE THERE, LL is EMPTY")

  def insertion(self,data):
    # 1st Time
    if self.head == None:      # if LL is empty Ist Node will be Head and Tail
      self.head = Node(data)   # Head will be pointed to Ist Node with data
      self.tail = self.head         # Tail will also point to 1 Node like Head
                          # 1 hi node hai, vahi head and vahi tail

    # Generic Nth Time
    else:
      new_node = Node(data)   # New Node Created with next = NULL and Data
      new_node.next=self.head      # Link from NewNode to Head
      self.head = new_node         # Head will point to NewNode

    print("insertion Successful with Data - ", data)

  def display(self):
    print("-----------------Display-----------------")
    if self.head == None:
      print("Empty Linked List, Nothing to Display")
    else:
      # Linked LIst is not Empty - Head to Tail Move and Print Data
      # Head and Tail wont change
      # Additional Marker will move from Head to Tail

      temp=self.head

      while(temp!=None):
        print(temp.data)
        temp=temp.next


    print("----------------------------------------")





```

```python
print("sll = Operations()")
sll = Operations()   # Object of Linked List Operation Class,
                      # that will do all operations and will store all data
print("sll.display()")
sll.display()

print("sll.insertion(11)")
sll.insertion(11)

print("sll.display()")
sll.display()

print("sll.insertion(22)")
sll.insertion(22)

print("sll.display()")
sll.display()

print("sll.insertion(33)")
sll.insertion(33)

print("sll.display()")
sll.display()






```
